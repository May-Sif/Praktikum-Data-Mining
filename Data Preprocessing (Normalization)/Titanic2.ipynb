```python
import pandas as pd
dataset = pd.read_csv('Titanic-Dataset.csv')
print(dataset)
```

```python
rows, cols = dataset.shape

print("Number of rows: ", rows)
print("Number of columns: ", cols)
```

```python
data = dataset.loc[:, ['Age', 'Fare']]
print (data)
```

```python
class_ = dataset['Survived']
print(class_)
```

```python
data['Age'] = dataset['Age'].fillna(data.groupby(class_)['Age'].transform('mean')) 

print('data setelah pengisian missing value pada kolom Age\n', data)
```

```python
# implementasi algoritma min-max

def min_max_normalization(data, new_min, new_max):
    minimum = data.min()
    maximum = data.max()

    return (data - minimum) * (new_max - new_min) / (maximum - minimum) + new_min

data['Age'] = min_max_normalization(data['Age'], 0, 1)
data['Fare'] = min_max_normalization(data['Fare'], 0, 1)

print(data)
```

```python
# implementasi algoritma Z-Score

def z_score_normalization(data):
    mean = data.mean()
    std_dev = data.std()

    new_data = (data - mean) / std_dev
    return new_data

data['Age'] = z_score_normalization(data['Age'])
data['Fare'] = z_score_normalization(data['Fare'])
print(data)
```

```python
# implementasi algoritma Sigmoidal
import numpy as np

def  sigmoidal_normalization(data):

    x = (data - data.mean()) / data.std()
    new_data = (1 - np.exp(-x) / (1 + np.exp(-x)))

    return new_data

data['Age'] = sigmoidal_normalization(data['Age'])
data['Fare'] = sigmoidal_normalization(data['Fare'])
print(data)
```

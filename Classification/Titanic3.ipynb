```python
import pandas as pd
dataset = pd.read_csv('titanic.csv')
print(dataset)
```

```python
test_dataset = pd.read_csv('titanic_test.csv')
print(test_dataset)
```

```python
train_data = dataset.loc[:, ['Age', 'Fare']]

# Catat posisi/index baris yang memiliki missing value
pos_missing_train = train_data[train_data.isnull().any(axis=1)].index.tolist()

print("\nPosisi missing train:")
print(pos_missing_train)

# Hapus baris yang memiliki missing value
train_data = train_data.dropna()
```

```python
test_data = test_dataset.loc[:, ['Age', 'Fare']]

# Catat posisi/index baris yang memiliki missing value
pos_missing_test = test_data[test_data.isnull().any(axis=1)].index.tolist()

print("\nPosisi missing test:")
print(pos_missing_test)

# Hapus baris yang memiliki missing value
test_data = test_data.dropna()
```

```python
train_label = dataset["Survived"].drop(index=pos_missing_train)

print("\nTrain label:")
print(train_label.head())
print("Jumlah train label:", len(train_label))
```

```python
test_label_dataset = pd.read_csv("titanic_testlabel.csv")

# Ambil kolom Survived sebagai label
test_label = test_label_dataset["Survived"]

# Hapus label pada posisi yang memiliki missing value
test_label = test_label.drop(index=pos_missing_test)

print("\nPengecekan jumlah data")
print("Train data  :", len(train_data))
print("Train label :", len(train_label))
print("Test data   :", len(test_data))
print("Test label  :", len(test_label))
```

```python
train_min = train_data.min()
train_max = train_data.max()

# Normalisasi min-max
def min_max_normalization(data, new_min, new_max):
    return (data - train_min) * (new_max - new_min) / (train_max - train_min) + new_min

train_data_normalized = min_max_normalization(train_data, 0, 1)
print("\nTrain data setelah normalisasi min-max:\n")
print(train_data_normalized)
```

```python
test_data = (test_data - train_min) * (1 - 0) / (train_max - train_min) + 0
print("\nTest data setelah normalisasi min-max:\n")
print(test_data)
```

```python
from sklearn.neighbors import KNeighborsClassifier
# Klasifikasi test_data kedalam train_data menggunakan KNN dengan k=1-15

class_results = {}

for k in range(1, 16):

    # Membuat model KNN
    knn = KNeighborsClassifier(n_neighbors=k)

    # Training
    knn.fit(train_data, train_label)

    # Prediksi test data
    class_result = knn.predict(test_data)

    # Precision / accuracy
    precision_ratio=knn.score(test_data, test_label)

    # error ratio
    error_ratio=1-precision_ratio

    # simpan hasil klasifikasi untuk k tertentu
    class_results[k] = class_result

class_result_df = pd.DataFrame(class_results)

print("\nClass Result:\n")
print(class_result_df.to_string(index=False))
print("Jumlah baris:", class_result_df.shape[0])
```

```python
results = []

for k in range(1, 16):

    # Ambil hasil klasifikasi dari nomor 9
    class_result = class_results[k]

    # Bandingkan dengan test_label
    error_count = (class_result != test_label.values).sum()

    # Hitung error ratio
    error_ratio = error_count / len(test_label)

    results.append({
        "k": k,
        "error_count": error_count,
        "error_ratio": error_ratio,
        "error_percentage": error_ratio * 100
    })

result_df = pd.DataFrame(results)

print("Hasil Error:")
print(result_df)
```

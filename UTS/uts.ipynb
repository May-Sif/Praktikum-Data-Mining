```python
import pandas as pd
import numpy as np

from sklearn.neighbors import KNeighborsRegressor, KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import confusion_matrix, classification_report
```

```python
dataset = pd.read_csv("titanic.csv")

print("Jumlah data:", len(dataset))
dataset.head()
```

```python
data = dataset[
    ["Sex", "Age", "Pclass", "Fare", "Survived"]
].copy()

data["Sex"] = data["Sex"].map({
    "male": 0,
    "female": 1
})

data.head()
```

```python
train_data = data.loc[
    data["Age"].notna(),
    ["Sex", "Pclass", "Fare", "Survived"]
].copy()

train_label = data.loc[
    data["Age"].notna(),
    "Age"
].copy()

test_data = data.loc[
    data["Age"].isna(),
    ["Sex", "Pclass", "Fare", "Survived"]
].copy()

print("Jumlah Train Data :", len(train_data))
print("Jumlah Train Label:", len(train_label))
print("Jumlah Test Data  :", len(test_data))
```

```python
scaler_age = MinMaxScaler()

train_data_normalized = scaler_age.fit_transform(train_data)

min_values = pd.Series(
    scaler_age.data_min_,
    index=train_data.columns
)

max_values = pd.Series(
    scaler_age.data_max_,
    index=train_data.columns
)

print("Nilai Minimum:")
print(min_values)

print("\nNilai Maximum:")
print(max_values)
```

```python
test_data_normalized = scaler_age.transform(test_data)

print("Hasil Normalisasi Test Data:")
print(test_data_normalized)
```

```python
knn_age = KNeighborsRegressor(n_neighbors=3)

knn_age.fit(train_data_normalized, train_label)

class_result = knn_age.predict(test_data_normalized)

print("Hasil Prediksi Age:")
print(class_result)
```

```python
data.loc[
    data["Age"].isna(),
    "Age"
] = class_result

print("Jumlah Missing Value Age:", data["Age"].isna().sum())

data.head()
```

```python
test_dataset = pd.read_csv("titanic_test.csv")

print("Jumlah Data Testing:", len(test_dataset))
test_dataset.head()
```

```python
train_data = data[
    ["Sex", "Age", "Pclass", "Fare"]
].copy()

train_label = data["Survived"].copy()

print("Jumlah Training Data :", len(train_data))
print("Jumlah Training Label:", len(train_label))

train_data.head()
```

```python
test_data = test_dataset[
    ["Sex", "Age", "Pclass", "Fare"]
].copy()

test_data["Sex"] = test_data["Sex"].map({
    "male": 0,
    "female": 1
})

test_data = test_data.dropna()

print("Jumlah Testing Data:", len(test_data))

test_data.head()
```

```python
test_label_dataset = pd.read_csv("titanic_testlabel.csv")

test_label = test_label_dataset.loc[
    test_data.index,
    "Survived"
].copy()

print("Jumlah Testing Label:", len(test_label))

test_label.head()
```

```python
scaler = MinMaxScaler()

train_data_normalized = scaler.fit_transform(train_data)

min_values = pd.Series(
    scaler.data_min_,
    index=train_data.columns
)

max_values = pd.Series(
    scaler.data_max_,
    index=train_data.columns
)

print("Nilai Minimum:")
print(min_values)

print("\nNilai Maximum:")
print(max_values)
```

```python
test_data_normalized = scaler.transform(test_data)

print("Hasil Normalisasi Test Data:")
print(test_data_normalized)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(train_data_normalized, train_label)

class_result = knn.predict(test_data_normalized)

print("Hasil Klasifikasi:")
print(class_result)
```

```python
error = np.sum(
    class_result != test_label.to_numpy()
)

print("Jumlah Error:", error)
```

```python
error_ratio = (error / len(test_data)) * 100

print(f"Error Ratio: {error_ratio:.2f}%")
```

```python
hasil_klasifikasi = pd.DataFrame({
    "Actual": test_label.to_numpy(),
    "Predicted": class_result
})

hasil_klasifikasi["Error"] = (
    hasil_klasifikasi["Actual"] != hasil_klasifikasi["Predicted"]
).astype(int)

hasil_klasifikasi["Status"] = np.where(
    hasil_klasifikasi["Error"] == 1,
    "Salah",
    "Benar"
)

hasil_klasifikasi
```

```python
total_data = len(hasil_klasifikasi)

total_error = hasil_klasifikasi["Error"].sum()

total_benar = total_data - total_error

print("Total Data Testing :", total_data)
print("Total Prediksi Benar:", total_benar)
print("Total Prediksi Salah:", total_error)
```

```python
accuracy = (total_benar / total_data) * 100

print(f"Accuracy: {accuracy:.2f}%")
```

```python
print("HASIL EVALUASI KNN")
print("Algoritma           : K-Nearest Neighbors")
print("Nilai K             : 3")
print("Jumlah Training Data:", len(train_data))
print("Jumlah Testing Data :", total_data)
print("Jumlah Benar        :", total_benar)
print("Jumlah Error        :", total_error)
print(f"Error Ratio         : {error_ratio:.2f}%")
print(f"Accuracy            : {accuracy:.2f}%")
```

```python
cm = confusion_matrix(
    test_label,
    class_result,
    labels=[0, 1]
)

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(
    classification_report(
        test_label,
        class_result,
        labels=[0, 1],
        zero_division=0
    )
)
```

```python
cm_df = pd.DataFrame(
    cm,
    index=["Actual 0", "Actual 1"],
    columns=["Predicted 0", "Predicted 1"]
)

cm_df
```

```python
hasil_klasifikasi.to_csv(
    "hasil_klasifikasi_titanic.csv",
    index=False
)

print("Hasil klasifikasi berhasil disimpan.")
```

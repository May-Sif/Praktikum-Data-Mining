```python
# Mengubah nama kolom menjadi lebih deskriptif

df = pd.read_csv("heart.csv", header=None)
df.columns = [
    "age",
    "sex",
    "cp",
    "trestbps",
    "chol",
    "fbs",
    "restecg",
    "thalach",
    "exang",
    "oldpeak",
    "slope",
    "ca",
    "thal",
    "Presence"
]

display(df.head())
```

```python
# Menghitung jumlah data untuk setiap kelas

print(df["Presence"].value_counts())
```

```python
# CEK TIPE DATA

print(df.info())
```

```python
# Check nilai unique pada setiap kolom fitur

for column in df.select_dtypes(include="object").columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
# Konversi fitur kategorikal menjadi numerik

df["sex"] = df["sex"].map({
    "female": 0,
    "male": 1
})

df["cp"] = df["cp"].map({
    "asymptomatic": 0,
    "non_anginal_pain": 1,
    "atypical_angina": 2,
    "typical_angina": 3
})

df["restecg"] = df["restecg"].map({
    "normal": 0,
    "ventricular_hypertrophy": 1,
    "wave_abnormality": 2
})

df["exang"] = df["exang"].map({
    "no": 0,
    "yes": 1
})

df["slope"] = df["slope"].map({
    "upsloping": 0,
    "flat": 1,
    "downsloping": 2
})

df["thal"] = df["thal"].map({
    "normal": 0,
    "fixed_defect": 1,
    "reversable_defect": 2
})

df["Presence"] = df["Presence"].map({
    "Absence": 0,
    "Presence": 1
})

print(df.dtypes)
```

```python
# Cek hasil konversi
display(df.head())

print("\nTipe data:")
print(df.dtypes)

print("\nMissing value setelah konversi:")
print(df.isnull().sum())
```

```python
# Memisahkan fitur dan target

X = df.drop(columns=["Presence"])
y = df["Presence"]

print("X:")
display(X.head())

print("\ny:")
display(y.head())
```

```python
# implementasi algoritma min-max

def min_max_normalization(data, new_min, new_max):
    minimum = data.min()
    maximum = data.max()

    return (data - minimum) * (new_max - new_min) / (maximum - minimum) + new_min

X_minmax = min_max_normalization(X, 0, 1)

print(X_minmax.head())
```

```python
# implementasi algoritma z-score

def z_score_normalization(data):
    mean = data.mean()
    std_dev = data.std()

    new_data = (data - mean) / std_dev
    return new_data

X_zscore = z_score_normalization(X)
print(X_zscore.head())
```

```python
# implementasi algoritma sigmoidal

def sigmoidal_normalization(data):

    x = (data - data.mean()) / data.std()
    new_data = (1 - np.exp(-x) / (1 + np.exp(-x)))

    return new_data

X_sigmoid = sigmoidal_normalization(X)
print(X_sigmoid.head())
```

```python
# Klasifikasi menggunakan k-NN (k = 1, k = 5, k = 9) dan decision tree (max_depth = 5)

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

models = {
    "k-NN (k=1)": KNeighborsClassifier(n_neighbors=1),
    "k-NN (k=5)": KNeighborsClassifier(n_neighbors=5),
    "k-NN (k=9)": KNeighborsClassifier(n_neighbors=9),
    "Decision Tree (max_depth=5)": DecisionTreeClassifier(max_depth=5, random_state=42),
}
```

```python
# Cross-validation menggunakan K-Fold (k = 4, k = 10) dan Leave-One-Out

from sklearn.model_selection import KFold, LeaveOneOut, cross_val_score

validations = {
    "4-Fold": KFold(n_splits=4, shuffle=True, random_state=42),
    "10-Fold": KFold(n_splits=10, shuffle=True, random_state=42),
    "Leave-One-Out": LeaveOneOut()
}
```

```python
results = []

normalizations = {
    "Min-Max": X_minmax,
    "Z-Score": X_zscore,
    "Sigmoidal": X_sigmoid
}

for norm_name, X_norm in normalizations.items():
    for model_name, model in models.items():
        for validation_name, validation in validations.items():
            scores = cross_val_score(
                model,
                X_norm,
                y,
                cv=validation,
                scoring="accuracy"
            )

            accuracy = scores.mean()
            error_ratio = 1 - accuracy
            results.append({
                "Normalisasi": norm_name,
                "Klasifikasi": model_name,
                "Validasi": validation_name,
                "Accuracy": accuracy,
                "Error Ratio": error_ratio
            })

results_df = pd.DataFrame(results)

print("Jumlah eksperimen:", len(results_df))
display(results_df)
```

```python
results_df["Accuracy (%)"] = results_df["Accuracy"] * 100
results_df["Error Ratio (%)"] = results_df["Error Ratio"] * 100

hasil = results_df[
    [
        "Normalisasi",
        "Klasifikasi",
        "Validasi",
        "Accuracy (%)",
        "Error Ratio (%)"
    ]
]

display(hasil)
```

```python
hasil_terbaik = hasil.sort_values(
    by="Error Ratio (%)"
).reset_index(drop=True)

display(hasil_terbaik)
```

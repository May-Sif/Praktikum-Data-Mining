```python
import pandas as pd
dataset = pd.read_csv('Titanic-Dataset.csv')
print(dataset)
```

```python
rows, cols = dataset.shape

print("Number of rows: ", rows)
print("Number of columns: ", cols)
```

```python
data = dataset.loc[:, ['Name', 'Sex', 'Age', 'Pclass', 'Fare']]
print (data)
```

```python
class_ = dataset['Survived']

print(class_)
```

```python
dataset["Relatives"] = dataset["SibSp"] + dataset["Parch"]

print(data["Relatives"])
```

```python
pclass_count = dataset['Pclass'].value_counts()

print(pclass_count)
```

```python
sex_count = dataset['Sex'].value_counts()

print(sex_count)
```

```python
survived_pclass = pd.crosstab(
    dataset["Pclass"],
    dataset["Survived"]
)

# Mengubah nama kolom 
survived_pclass.columns = ["Tidak Selamat (0)", "Selamat (1)"]

print(survived_pclass)
```

```python
dataset.plot(
    x="PassengerId",
    y="Sex",
    kind="scatter",
    c="Survived",
    colormap="viridis",
    title="Visualisasi Data Titanic (sex)"
)
```

```python
dataset.plot(
    x="PassengerId",
    y="Age",
    kind="scatter",
    c="Survived",
    colormap="viridis",
    title="Visualisasi Data Titanic (age)"
)
```

```python
import matplotlib.pyplot as plt
plt.scatter(dataset["PassengerId"], dataset["Age"], c=dataset["Age"], cmap="Paired")
plt.xlabel("Passenger ID")
plt.ylabel("Age")
plt.title("Age Distribution of Passengers")
plt.show()
```
